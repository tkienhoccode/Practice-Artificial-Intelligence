    'K': ['B', 'C', 'I', 'L'],
    'L': ['A', 'C', 'K'],
    'M': ['D', 'E', 'F', 'G', 'I', 'J']
}

# Tính bậc
bac = {}

for vung in graph:
    bac[vung] = len(graph[vung])

print("Bậc các vùng:")
for vung in bac:
    print(vung, ":", bac[vung])

# Selection Sort giảm dần theo bậc
danh_sach = list(bac.items())

for i in range(len(danh_sach)):
    max_idx = i

    for j in range(i + 1, len(danh_sach)):
        if danh_sach[j][1] > danh_sach[max_idx][1]:
            max_idx = j

    danh_sach[i], danh_sach[max_idx] = danh_sach[max_idx], danh_sach[i]

thu_tu = [x[0] for x in danh_sach]

print("\nThứ tự tô màu:")
print(thu_tu)

````python
CODE PYTHON — K-MEANS
import numpy as np
from scipy.spatial.distance import cdist

X = np.array([
    [0,1],
    [4,3],
    [2,5],
    [1,9],
    [3,7],
    [2,10]
])

centers = np.array([
    [2,2],
    [2,8]
])

distances = cdist(X, centers)

print("Ma trận khoảng cách:")
print(distances)

labels = np.argmin(distances, axis=1)

print("\nNhãn cụm:")
print(labels)

new_centers = np.array([
    X[labels == i].mean(axis=0)
    for i in range(2)
])

print("\nTâm mới:")
print(new_centers)
````

````python
# CODE TOÀN BỘ CÂU 2 — K-MEANS & K-NN

```python
# =========================
# CÂU 2A — K-MEANS
# =========================

import numpy as np
from scipy.spatial.distance import cdist

# Danh sách điểm
X = np.array([
    [0, 1],   # P1
    [4, 3],   # P2
    [2, 5],   # P3
    [1, 9],   # P4
    [3, 7],   # P5
    [2, 10]   # P6
])

# Tâm khởi tạo
C = np.array([
    [2, 2],   # C1
    [2, 8]    # C2
])

print("===== K-MEANS =====\n")

# Tính khoảng cách Euclid
D = cdist(X, C)

print("Ma trận khoảng cách:\n")
print(D)

# Gán nhãn cụm
labels = np.argmin(D, axis=1)

print("\nNhãn cụm:")

for i in range(len(labels)):
    print("P", i+1, "-> Cụm", labels[i] + 1)

# Cập nhật tâm mới
new_centers = np.array([
    X[labels == i].mean(axis=0)
    for i in range(2)
])

print("\nTâm mới:\n")
print(new_centers)

# Kiểm tra hội tụ
if np.array_equal(C, new_centers):
    print("\nThuật toán đã hội tụ")
else:
    print("\nThuật toán chưa hội tụ")


# =========================
# CÂU 2B — KNN
# =========================

print("\n=========================")
print("===== KNN =====")
print("=========================\n")

# Tập huấn luyện
train = [
    ((1,1), 1),  # T1
    ((2,1), 0),  # T2
    ((2,2), 1),  # T3
    ((1,3), 0),  # T4
    ((2,3), 0),  # T5
    ((3,3), 1)   # T6
]

# Điểm cần phân loại
q = (0,0)

distances = []

# Tính khoảng cách
for point, label in train:

    d = (point[0] - q[0])**2 + (point[1] - q[1])**2

    distances.append((point, label, d))

# Sắp xếp tăng dần
distances.sort(key=lambda x: x[2])

print("Khoảng cách tăng dần:\n")

for item in distances:
    print(item)

# =========================
# K = 3
# =========================

k = 3

classes = {}

for i in range(k):

    label = distances[i][1]

    if label in classes:
        classes[label] += 1
    else:
        classes[label] = 1

result = max(classes, key=classes.get)

print("\nK = 3")
print("Kết quả phân loại:", result)

# =========================
# K = 4
# =========================

k = 4

classes = {}

for i in range(k):

    label = distances[i][1]

    if label in classes:
        classes[label] += 1
    else:
        classes[label] = 1

result = max(classes, key=classes.get)

print("\nK = 4")
print("Kết quả phân loại:", result)

# =========================
# K = 5
# =========================

k = 5

classes = {}

for i in range(k):

    label = distances[i][1]

    if label in classes:
        classes[label] += 1
    else:
        classes[label] = 1

result = max(classes, key=classes.get)

print("\nK = 5")
print("Kết quả phân loại:", result)
````
